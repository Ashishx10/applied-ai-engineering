```python
# Loading Data From a file -> In NumPy, data from text files (such as `.txt` or `.csv` files) can be 
# loaded into arrays using the **np.genfromtxt()** function
import numpy as np
np.genfromtxt('test.txt',delimiter=',') # Auto casted into float type
```

```python
# Converting float type into int type -> .astype('int...')
import numpy as np
loaded = np.genfromtxt('test.txt',delimiter = ',')
print(loaded)
typecast = loaded.astype('int32')
print(typecast) 
```

```python
# Boolean Masking(Filtering Data) And Advanced indexing
import numpy as np
loaded >10
loaded[loaded<7]
```
