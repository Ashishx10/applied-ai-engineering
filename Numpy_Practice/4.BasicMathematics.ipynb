```python
# Basic Mathematics(Algebra, Trigonometry)
# Scalar Arithmetic
import numpy as np
a = np.array([1,2,3,4])
print(a+1) # addition(+)
print(a-1) # subtraction(-)
print(a*3) # multiplication(*)
print(a/2) # division(/)
print(a**2) # power(**)
```

```python
# Elementwise operation
import numpy as np
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
print(a+b) # addition(+)
print(a-b) # subtraction(-)
print(a*b) # multiplication(*)
print(a/b) # division(/)
print(a**b) # power(**)
```

```python
# Vectorised Math Functions and Constants
import numpy as np
r = 3
a = np.array([1,2,3,4])
print(np.sqrt(a)) # square root -> np.sqrt(arr_var)
b = np.array([1,2,3,4])
print((np.sin(b),(np.cos(b)))) # Trigonometric functions ->np.sin/other functions(arr_var)
c = np.array([1.09,2.99,3.11,9.99])
print(np.round(c)) # round down -> np.round/floor()
d = np.array([1.09,2.99,3.11,9.99])
print(np.ceil(c)) # round up -> np.ceil()
print(np.pi * (r**2)) # Constants -> np.pi

```

```python
# Linear Algbera and Matrix
import numpy as np
a = np.ones((2,3))
print(a)
b = np.full((3,2),2)
print(b)
print(np.matmul(a,b)) # matrix multiplication -> np.matmul(arr1,arr2)
c = np.identity(3)
print(np.linalg.det(c)) # Matrix determinant -> np.linalg.det(arr)

```

```python
# Aggregated Mathematical functions
import numpy as np
a = np.array([10,20,30,40,50])
print(np.sum(a)) # Sum -> np.sum(arr)
print(np.average(a)) # Average/mean -> np.average/mean(arr)
print(np.min(a)) # Min value -> np.min(arr)
print(np.max(a)) # Max value -> np.max(arr)
print(np.std(a)) # Standard deviation -> np.std(arr)
print(np.var(a)) # variance -> np.var(arr)
```

```python
# Compute the dot product of two vectors, and matmul of a 2×3 and a 3×2.
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])
vector_dot = np.dot(v1, v2)
m1 = np.array([[1, 2, 3], 
               [4, 5, 6]])
m2 = np.array([[7, 8],
               [9,10],
               [11, 12]])
matrix_matmul = np.matmul(m1, m2)
print("Vector Dot Product Target Result:", vector_dot)
print("\nMatrix Multiplication Result:\n", matrix_matmul)
```
