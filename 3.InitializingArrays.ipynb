```python
# Initializing Different Types of Arrays
```

```python
# All 0's Matrix -> np.zeros([shape])
import numpy as np
a = np.zeros(5) #1D
b = np.zeros([3,3])
c = np.zeros([2,4,4],dtype='int32')
print(a)
print(b)
print(c)
```

```python
# All 1's Matrix -> np.ones([shape])
import numpy as np
a = np.ones(5) #1D
b = np.ones([3,3])
c = np.ones([2,4,4],dtype='int32')
print(a)
print(b)
print(c)
```

```python
# Any other number -> np.full((shape),value to be filled)
import numpy as np
a = np.full((5),20)
b = np.full((2,2),15)
c = np.full((1,2,2),10,dtype='int32')
print(a)
print(b)
print(c)
```

```python
# copying shape -> np.full_like(shape of array, value to be filled)
import numpy as np
a = np.array([[1,2,3,4],[5,6,7,8]])
b = np.full_like(a,10)
print(a)
print(b)
```

```python
# Random Decimal Number Arrays -> np.random.rand(shape) or 
# rng = np.random.default_rng() {modern Generator API } // have to pass shape as tuple
#       rng.random
import numpy as np
rng = np.random.default_rng()
a = np.random.rand(5)
b = rng.random((2,2))
c = rng.random((2,2,2))
d = np.random.rand(2,2)
print(a)
print(b)
print(c)
print(d)
```

```python
# Random Integer Number Arrays -> np.random.randint(low,high,size) or 
# rng = np.random.default_rng(low,high{inclusive},size) // have to pass shape as tuple
#        rng.integers
import numpy as np
rng = np.random.default_rng()
a = rng.integers(1,7,size = 6)
b = np.random.randint(1,5,size = (2,2))
c = rng.integers(1,9,size = (2,2,2))
print(a)
print(b)
print(c)
```

```python
# Special Strutural Arrays
# Identity Matrix -> np.identity(shape)
import numpy as np
a = np.identity(5)
print(a)
```

```python
# Repeating Arrays -> np.repeat(arr_var,times,axis)
import numpy as np
a = np.array([[1,2,3,4]])
a1 = np.repeat(a,3,axis = 0)
print(a)
print(a1)

```

```python
# Initializing a large matrix
import numpy as np
a = np.ones((5,5)) #creating a 5*5 matrix filled with 1's
print(a)
b = np.zeros((3,3)) #creating a 3*3 matrix filled with 0's
b[1,1] = 9 # replacing the [row_idx=1,col_idx=1] with 9
print(b)
a[1:4,1:4] = b # replacing the [row_idx(1:4),col_idx(1:3)] with b to fill the middle values
print(a)
```

```python
# Copying Arrays safely -> .copy()
# Assigning an existing array to a new variable via `b = a` creates a pointer reference rather than a 
# new array. Modifying `b` will accidentally change the contents of `a`.To create an independent duplicate
# always use copy():
import numpy as np
a = np.array([1,2,3])
#b = a // if .copy() is not used 
b = a.copy()
b[0] = 10
print(b) # if .copy() not used b -> 10,2,3
print(a)  #if .copy() not used a -> 10,2,3
```
